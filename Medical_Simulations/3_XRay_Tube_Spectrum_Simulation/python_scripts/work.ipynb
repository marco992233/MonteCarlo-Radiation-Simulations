## No filter - electrons @ 150keV


```python
# 1. Generazione dello spettro (Caso 1: Input Originale)
# kvp = 150, angolo (th) = 45 gradi, target = 'W' (Tungsteno)
caso1 = sp.Spek(kvp=150, th=45, targ='W')

# 3. Estrazione dei dati per il grafico
energia, fluenza = caso1.get_spectrum(edges=False)

# 4. Creazione del grafico
plt.figure(figsize=(8, 5))
plt.plot(energia, fluenza, color='blue', label='W, 150 kVp, 45°')

# Formattazione del grafico
plt.title('Caso 1: Spettro Raggi X (Input Originale)')
plt.xlabel('Energia dei fotoni (keV)')
plt.ylabel('Fluenza (fotoni / cm$^2$ / mAs)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()
plt.xlim(0, 160) # Impostiamo il limite poco dopo i 150 keV
plt.savefig('W_150_45_nofilter.png', dpi=200, bbox_inches='tight')
plt.show()
```

## Al filter 1mm - electrons @ 150keV


```python
# 1. Generazione dello spettro
# W = Tungsteno, 150 kVp, 45 gradi
s = sp.Spek(kvp=150, th=45, targ='W')
s.filter('Al', 1.0) # 1 mm di Alluminio

# Estrazione dati (energia in keV)
energia, fluenza = s.get_spectrum(edges=False)

# 2. Normalizzazione dell'area a 1
# Dividiamo la fluenza per l'integrale della curva
fluenza_norm = fluenza / np.trapezoid(fluenza, energia)

# 3. Creazione del grafico
plt.figure(figsize=(8, 5))
plt.plot(energia, fluenza_norm, color='blue', lw=2, label='SpekPy Normalizzato (Area=1)')

# Formattazione
plt.title('Spettro SpekPy Normalizzato\n(W, 150 kVp, 45°, 1 mm Al)')
plt.xlabel('Energia (keV)')
plt.ylabel('Densità di probabilità (Area=1)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.xlim(0, 160)
plt.ylim(0, max(fluenza_norm) * 1.1) # Dà respiro ai picchi K

plt.savefig('W_150_45_1mm_spekpy_norm.png', dpi=200, bbox_inches='tight')
plt.show()
```

## Comparison w/ Penelope: electrons @ 150keV + Al filter 1mm (ok)


```python
# --- 1. Generazione spettro SpekPy ---
# Configurazione: Tungsteno, 150 kVp, 45 gradi, 1 mm Al
s = sp.Spek(kvp=150, targ='W', th=45)
s.filter('Al', 1.0) 
energy_spek, flux_spek = s.get_spectrum(edges=False, flu=True)

# Normalizzazione SpekPy (Area = 1)
area_spek = np.trapezoid(flux_spek, energy_spek)
flux_spek_norm = flux_spek / area_spek

# --- 2. Caricamento dati Penelope ---
path_pen = '/Users/marcofumagalli/Desktop/Erasmus2026/Courses/Aplicaciones Médicas e Industriales de las radiaciones/Simulations/Tubo X-ray simulation/Simulation_Penelope/My_penelope/fsource/5-x-ray-tube/spc-impdet-02.dat'

# Carichiamo i dati: Col 0 = Energia (eV), Col 1 = Intensità, Col 2 = Incertezza (1 sigma)
data_pen = np.loadtxt(path_pen)
energy_pen = data_pen[:, 0] / 1000.0  # eV -> keV
flux_pen = data_pen[:, 5]
uncertainty_pen = data_pen[:, 6]      # Sexta colonna del file .dat di Penelope

# Normalizzazione Penelope (Area = 1)
area_pen = np.trapezoid(flux_pen, energy_pen)
flux_pen_norm = flux_pen / area_pen
error_pen_norm = uncertainty_pen / area_pen # Riscaliamo anche l'errore

# --- 3. Plot di confronto con OMBREGGIATURA ---
plt.figure(figsize=(10, 6))

# Plot SpekPy (Analitico) - Linea nera continua per contrasto massimo
plt.plot(energy_spek, flux_spek_norm, label='SpekPy (Analitico)', color='black', lw=1.5, zorder=1)

# Plot Penelope (Monte Carlo)
# 1. Linea principale a gradini
plt.step(energy_pen, flux_pen_norm, where='mid', label='Penelope (Monte Carlo)', 
         color='blue', lw=1.2, zorder=3)

# 2. Incertezza ombreggiata (Shaded area)
y_upper = flux_pen_norm + error_pen_norm
y_lower = flux_pen_norm - error_pen_norm

plt.fill_between(energy_pen, y_lower, y_upper, step='mid', 
                 color='blue', alpha=0.2, label=r'Incertezza Penelope ($3\sigma$)', zorder=2)

# Formattazione estetica
plt.title('Confronto Spettri Normalizzati: Penelope vs SpekPy\n(W, 150 kVp, 45°, 1 mm Al)')
plt.xlabel('Energia (keV)')
plt.ylabel('Densità di Probabilità (1/keV)')
plt.legend(loc='upper right', fontsize='small')
plt.grid(True, which='both', linestyle=':', alpha=0.6)

# Limiti assi
plt.xlim(0, 160)
max_val = max(np.max(flux_spek_norm), np.max(flux_pen_norm))
plt.ylim(0, max_val * 1.1)

plt.tight_layout()
plt.savefig('fig02_W_150_45_1mm.png', dpi=200, bbox_inches='tight')
plt.show()
```

## Comparison w/ Penelope: electrons @ 150keV + Al filter 2mm + 20deg (ok)


```python
# --- 1. Generazione spettro SpekPy ---
s = sp.Spek(kvp=150, targ='W', th=20)
s.filter('Al', 2.0) 
energy_spek, flux_spek = s.get_spectrum(edges=False, flu=True)

area_spek = np.trapezoid(flux_spek, energy_spek)
flux_spek_norm = flux_spek / area_spek

# --- 2. Caricamento dati Penelope ---
path_pen = '/Users/marcofumagalli/Desktop/Erasmus2026/Courses/Aplicaciones Médicas e Industriales de las radiaciones/Simulations/Tubo X-ray simulation/Simulation_Penelope/My_penelope/fsource/5-x-ray-tube_150_20_2mm/spc-impdet-02.dat'

try:
    data_pen = np.loadtxt(path_pen)
    energy_pen_raw = data_pen[:, 0] / 1000.0
    flux_pen_raw = data_pen[:, 5]
    uncertainty_pen_raw = data_pen[:, 6]
    # Maschera per pulire gli zeri sentinella (1e-35) di PENELOPE
    mask = flux_pen_raw > 1e-34
    energy_pen = energy_pen_raw[mask]
    flux_pen = flux_pen_raw[mask]
    uncertainty_pen = uncertainty_pen_raw[mask]

    area_pen = np.trapezoid(flux_pen, energy_pen)
    flux_pen_norm = flux_pen / area_pen
    error_pen_norm = uncertainty_pen / area_pen
    
except Exception as e:
    print(f"Errore: {e}")
    energy_pen, flux_pen_norm, error_pen_norm = None, None, None

# --- 3. Plot di confronto con OMBREGGIATURA ---
plt.figure(figsize=(10, 6))

# Plot SpekPy (Analitico)
plt.plot(energy_spek, flux_spek_norm, label='SpekPy (W, 20°, 2mm Al)', 
         color='black', lw=1.5, zorder=1)

if energy_pen is not None:
    # 1. Linea principale Penelope
    plt.step(energy_pen, flux_pen_norm, where='mid', label='Penelope (Monte Carlo)', 
             color='red', lw=1.2, zorder=3)
    
    # 2. VISUALIZZAZIONE ERRORE: Fascia ombreggiata (fill_between)
    # Calcoliamo il limite superiore e inferiore dell'errore
    y_upper = flux_pen_norm + error_pen_norm
    y_lower = flux_pen_norm - error_pen_norm
    
    # Usiamo step per creare l'ombra che segue i gradini
    plt.fill_between(energy_pen, y_lower, y_upper, step='mid', 
                     color='red', alpha=0.2, label=r'Incertezza Penelope ($3\sigma$)', zorder=2)

# Formattazione
plt.title('Confronto Spettri: Visualizzazione con Incertezza Ombreggiata')
plt.xlabel('Energia (keV)')
plt.ylabel('Intensità Normalizzata (Area=1)')
plt.legend(loc='upper right')
plt.grid(True, linestyle=':', alpha=0.6)
plt.xlim(0, 160)

if energy_pen is not None:
    plt.ylim(0, np.max(flux_spek_norm) * 1.1)

plt.tight_layout()
plt.savefig('fig05_W_150_20_2mm.png', dpi=200, bbox_inches='tight')
plt.show()
```

```python
# --- 1. Generazione spettro SpekPy (Molibdeno) ---
# Parametri: Mo, 28 kVp, 20 gradi, 2.0 mm Al
s = sp.Spek(kvp=28, targ='Mo', th=20)
s.filter('Al', 2.0) 

# CORREZIONE: Aggiunto flu=True per confrontare la fluenza (fisica coerente con Penelope)
energy_spek, flux_spek = s.get_spectrum(edges=False, flu=True)

# Normalizzazione SpekPy (Area = 1)
area_spek = np.trapezoid(flux_spek, energy_spek)
flux_spek_norm = flux_spek / area_spek

# --- 2. Caricamento dati Penelope ---
path_pen = '/Users/marcofumagalli/Desktop/Erasmus2026/Courses/Aplicaciones Médicas e Industriales de las radiaciones/Simulations/Tubo X-ray simulation/Simulation_Penelope/My_penelope/fsource/5-x-ray-tube_Mo_28_20_2mm/spc-impdet-02.dat'

try:
    # Caricamento con gestione commenti
    data_pen = np.loadtxt(path_pen, comments='#')
    energy_pen_raw = data_pen[:, 0] / 1000.0  # eV -> keV
    
    # CORREZIONE: Colonne corrette (5: Fluenza, 6: Incertezza 3-sigma)
    flux_pen_raw = data_pen[:, 5]             
    uncertainty_pen_raw = data_pen[:, 6]      

    # CORREZIONE: Maschera per pulire il rumore di fondo (zero statistico < 1e-34)
    mask = flux_pen_raw > 1e-34
    energy_pen = energy_pen_raw[mask]
    flux_pen = flux_pen_raw[mask]
    uncertainty_pen = uncertainty_pen_raw[mask]

    # Normalizzazione Penelope (Area = 1)
    area_pen = np.trapezoid(flux_pen, energy_pen)
    flux_pen_norm = flux_pen / area_pen
    error_pen_norm = uncertainty_pen / area_pen
    
except Exception as e:
    print(f"Errore nel caricamento del file Penelope: {e}")
    energy_pen, flux_pen_norm, error_pen_norm = None, None, None

# --- 3. Plot di confronto con OMBREGGIATURA ---
plt.figure(figsize=(10, 6))

# Plot SpekPy (Analitico) - Nero continuo per risaltare i picchi Mo
plt.plot(energy_spek, flux_spek_norm, label='SpekPy (Mo, 28 kVp, 20°, 2mm Al)', 
         color='black', lw=1.5, zorder=1)

if energy_pen is not None:
    # 1. Linea principale Penelope (Rossa a gradini)
    plt.step(energy_pen, flux_pen_norm, where='mid', label='Penelope (Monte Carlo)', 
             color='red', lw=1.2, zorder=3)
    
    # 2. Fascia ombreggiata dell'errore (Nuvola)
    y_upper = flux_pen_norm + error_pen_norm
    y_lower = np.maximum(0, flux_pen_norm - error_pen_norm) # Evita valori negativi grafici
    
    plt.fill_between(energy_pen, y_lower, y_upper, step='mid', 
                     color='red', alpha=0.2, label=r'Incertezza Penelope ($3\sigma$)', zorder=2)

# Formattazione estetica per Mammografia
plt.title('Confronto Spettri: Mo a 28 kVp\nAnodo Molibdeno 20° - Filtro 2mm Al', fontsize=13)
plt.xlabel('Energia (keV)')
plt.ylabel('Intensità Normalizzata (Area=1)')
plt.legend(loc='upper right', fontsize='small')
plt.grid(True, linestyle=':', alpha=0.6)

# Limiti asse X per focalizzarsi sul range Mo (0-30 keV)
plt.xlim(0, 32) 

if energy_pen is not None:
    # Calcola il massimo tra i due spettri per non tagliare nulla
    max_y_real = max(np.max(flux_spek_norm), np.max(flux_pen_norm))
    plt.ylim(0, max_y_real * 1.1)

plt.tight_layout()
plt.savefig('fig07_Mo_28_20_2mm.png', dpi=200, bbox_inches='tight')
plt.show()
```
